# Gemini suggested Fun Webcam game

```python
import cv2
import time

# Open webcam or video file
cap = cv2.VideoCapture(0)

mode = 'normal'      # Options: 'normal', 'gray', 'edges', 'hsv'
delay = 1            # Default wait delay in ms (1ms = maximum speed)
paused = False       # Playback pause state

print("""
--- CONTROLS ---
'n' : Normal Mode
'g' : Grayscale Mode
'e' : Canny Edge Mode
'h' : HSV Color Mode
'f' : Fast Forward (Decrease delay)
's' : Slow Motion (Increase delay)
'p' : Pause / Resume Toggle
'q' : Quit Program
""")

while cap.isOpened():
    if not paused:
        ret, frame = cap.read()
        if not ret:
            break
        
        # Mirror for natural interaction
        frame = cv2.flip(frame, 1)

        # Apply selected filter based on key state
        if mode == 'gray':
            processed = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        elif mode == 'edges':
            gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
            processed = cv2.Canny(gray, 100, 200)
        elif mode == 'hsv':
            processed = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
        else:
            processed = frame.copy()

        # Render current HUD info on frame
        status_text = f"Filter: {mode.upper()} | Delay: {delay}ms | {'PAUSED' if paused else 'PLAYING'}"
        cv2.putText(processed, status_text, (20, 40), 
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

        cv2.imshow('Interactive Controls', processed)

    # Listen for key presses during the waitKey window
    key = cv2.waitKey(0 if paused else delay) & 0xFF

    # Key Handling Logic
    if key == ord('q'):
        break
    elif key == ord('n'):
        mode = 'normal'
    elif key == ord('g'):
        mode = 'gray'
    elif key == ord('e'):
        mode = 'edges'
    elif key == ord('h'):
        mode = 'hsv'
    elif key == ord('p'):
        paused = not paused
    elif key == ord('f'):
        delay = max(1, delay - 10)  # Faster playback
    elif key == ord('s'):
        delay += 10                 # Slower playback

cap.release()
cv2.destroyAllWindows()
```

```python
cap = cv2.VideoCapture(0)
mirror = True

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

codec = cv2.VideoWriter_fourcc('X','V','I','D')
#codec = cv2.VideoWriter_fourcc(*'XVID')
recorder = cv2.VideoWriter("First video with Open CV.avi", codec, 30, (width, height))

while True:
    bool, frame = cap.read()
    if not bool:
        break

    if mirror:
        processed = cv2.flip(frame, 1)
    else:
        processed = frame.copy()

    status_text = f'Mirrored : {"Yes" if mirror else "No"} | Delay : {1}'
    cv2.putText(processed, status_text, (20, 20), cv2.FONT_HERSHEY_COMPLEX, 0.9, (255, 255, 10), 1)
    
    recorder.write(processed)
    cv2.imshow("Webcam", processed)
    
    key = cv2.waitKey(1) & 0xFF

    if key == ord('m'):
        mirror = not mirror
    elif key == ord('q'):
        break

cap.release()
recorder.release()
cv2.destroyAllWindows()
```
