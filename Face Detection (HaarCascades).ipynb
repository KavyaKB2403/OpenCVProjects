```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

## Face Detection

```python
face_cascade = cv2.CascadeClassifier("haarcascade_frontalface_default.xml")
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        print("Error!! Cannot capture")
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    face = face_cascade.detectMultiScale(gray, 1.1, 8)

    for (x, y, w, h) in face:
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
        
    cv2.imshow("Face Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("Quiting")
        break

cap.release()
cv2.destroyAllWindows()
```

## Face, eye, smile Detection

```python
# Use cv2.data.haarcascades to load built-in XML files reliably
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
eye_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_eye.xml")
smile_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_smile.xml")

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        print("Error!! Cannot capture")
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Detect faces
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=6)

    # Everything operating on (x, y, w, h) MUST stay inside this loop
    for (x, y, w, h) in faces:
        # 1. Draw bounding box around detected face
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

        # 2. Extract Region of Interest (ROI) for THIS specific face
        roi_gray = gray[y:y + h, x:x + w]
        roi_frame = frame[y:y + h, x:x + w]

        # 3. Detect eyes inside the face ROI
        eyes = eye_cascade.detectMultiScale(roi_gray, scaleFactor=1.1, minNeighbors=6)
        if len(eyes) > 0:
            cv2.putText(frame, "Eye Detected", (x, y - 30), 
                        cv2.FONT_HERSHEY_COMPLEX, 0.5, (0, 0, 255), 2)

        # 4. Detect smile inside the face ROI (minNeighbors set higher to prevent false triggers)
        smiles = smile_cascade.detectMultiScale(roi_gray, scaleFactor=1.7, minNeighbors=8)
        if len(smiles) > 0:
            cv2.putText(frame, "Smile Detected", (x, y - 10), 
                        cv2.FONT_HERSHEY_COMPLEX, 0.5, (0, 0, 255), 2)

    cv2.imshow("Face Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("Quitting")q
        break

cap.release()
cv2.destroyAllWindows()
```

```python
eyeglasses = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_eye_tree_eyeglasses.xml")

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame = cv2.flip(frame, 1)
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    glasses = eyeglasses.detectMultiScale(gray, 1.1, 8)

    for (x, y, w, h) in glasses:
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 4)

    cv2.imshow("Eye Glasses Detection", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("Quiting")
        break

cap.release()
cv2.destroyAllWindows()
```

```python
upperbody = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_upperbody.xml"
)

cap = cv2.VideoCapture(0)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame = cv2.flip(frame, 1)

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    upbody = upperbody.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5
    )

    for (x, y, w, h) in upbody:
        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            4
        )

    cv2.imshow("Upper Body Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
